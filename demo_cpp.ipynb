```cpp
//2.3
#include <iostream>

// 1. Hàm VOID: Được gọi vì HÀNH VI (in ấn, xuất thông điệp), KHÔNG trả về dữ liệu
void printWelcomeBanner()
{
    std::cout << "=================================\n";
    std::cout << "   CHUONG TRINH TINH TOAN DEMO   \n";
    std::cout << "=================================\n";

    // BEST PRACTICE: Không viết "return;" ở cuối hàm void vì nó hoàn toàn thừa thãi.
    // Hàm sẽ tự động quay về Caller khi chạm dấu ngoặc nhọn kết thúc '}' dưới đây.
}

// 2. Hàm NON-VOID (có giá trị trả về): Dùng khi cần GỬI DỮ LIỆU về cho Caller
int calculateBonus()
{
    return 50; // Bắt buộc phải có giá trị trả về
}

// 3. Hàm VOID thông báo kết thúc
void printGoodbye()
{
    std::cout << "Tam biet va hen gap lai!\n";
}

int main()
{
    // -------------------------------------------------------------
    // NGỮ CẢNH 1: Độc lập (Không đòi hỏi giá trị)
    // Cả hàm void lẫn hàm trả về giá trị đều gọi được tại đây!
    // -------------------------------------------------------------
    printWelcomeBanner(); // Hợp lệ: Gọi hàm void để thực hiện hành vi

    calculateBonus();     // Hợp lệ: Hàm trả về 50, nhưng Caller không dùng
                          // nên số 50 này bị lờ đi và hủy bỏ (discarded).
                          // Lưu ý: Giá trị 50 KHÔNG tự in ra màn hình!

    // -------------------------------------------------------------
    // NGỮ CẢNH 2: Đòi hỏi giá trị (Require a value)
    // CHỈ hàm có giá trị trả về mới được xuất hiện tại đây!
    // -------------------------------------------------------------
    int totalScore{ 100 + calculateBonus() }; // Hợp lệ: 100 + 50 = 150
    std::cout << "Tong diem dat duoc: " << totalScore << '\n';

    // -------------------------------------------------------------
    // CÁC BẪY BIÊN DỊCH CẦN TRÁNH (COMPILE ERRORS):
    // -------------------------------------------------------------
    // std::cout << printWelcomeBanner(); 
    // -> LỖI: cout đòi hỏi giá trị cụ thể, không nhận 'void' (hư không)!

    // int errorVar{ printWelcomeBanner() }; 
    // -> LỖI: Không thể khởi tạo biến int từ một hàm void!

    printGoodbye(); // Kết thúc chương trình gọn gàng bằng hàm void

    return 0; // main() là hàm kiểu int, trả về Status code 0 cho Hệ điều hành
}
```

```cpp
//2.4
#include <iostream>

// =============================================================================
// 1. PARAMETERS (Tham số): Biến khai báo ở Function Header đón nhận dữ liệu
// =============================================================================
int add(int x, int y) // 'x' và 'y' là PARAMETERS (2 tham số)
{
    return x + y;
}

int multiply(int a, int b) // 'a' và 'b' là PARAMETERS
{
    return a * b;
}

// =============================================================================
// 2. CHỨNG MINH CƠ CHẾ "PASS BY VALUE" (Truyền theo giá trị)
// =============================================================================
void modifyValue(int p) // 'p' nhận một bản sao (copy) từ đối số truyền vào
{
    p = p * 10; // Thay đổi 'p' bên trong hàm này
    std::cout << "[Trong modifyValue] p = " << p << '\n';
    // Khi hàm kết thúc, biến tham số 'p' bị HỦY khỏi RAM
}

// =============================================================================
// 3. UNNAMED PARAMETER (Tham số không tên): Tắt cảnh báo -Wunused-parameter
// =============================================================================
// Khi hàm có tham số nhưng không dùng trong thân hàm, ta XÓA TÊN, chỉ giữ kiểu:
void logEvent(int eventId, int /*timestamp*/) // 'timestamp' được comment lại
{
    std::cout << "Ghi log Event ID: " << eventId << '\n';
    // Compiler sẽ KHÔNG phát cảnh báo 'unused parameter'
}

int main()
{
    // =========================================================================
    // 4. ARGUMENTS (Đối số): Giá trị / Biểu thức cụ thể gửi sang cho hàm
    // =========================================================================
    int result1{ add(3, 4) }; // '3' và '4' là ARGUMENTS (literals)
    std::cout << "3 + 4 = " << result1 << '\n';

    // Đối số có thể là bất kỳ BIỂU THỨC (Expression) hợp lệ nào:
    // Biểu thức (1 + 2) tính ra 3 -> copy vào x
    // Biểu thức (3 * 4) tính ra 12 -> copy vào y
    std::cout << "add(1 + 2, 3 * 4) = " << add(1 + 2, 3 * 4) << '\n'; // In ra 15

    // =========================================================================
    // 5. FUNCTION COMPOSITION: Lồng hàm con làm đối số cho hàm khác
    // =========================================================================
    // CPU giải quyết hàm con bên trong trước: multiply(2, 3) = 6
    // Sau đó chạy tiếp: add(1, 6) = 7
    std::cout << "add(1, multiply(2, 3)) = " << add(1, multiply(2, 3)) << '\n'; // In ra 7

    // =========================================================================
    // 6. KIỂM CHỨNG PASS BY VALUE: Biến gốc ở Caller KHÔNG bị ảnh hưởng
    // =========================================================================
    int original{ 5 };
    std::cout << "[main - Truoc khi goi ham] original = " << original << '\n'; // 5

    modifyValue(original); // Copy giá trị 5 của original sang biến 'p'

    std::cout << "[main - Sau khi goi ham] original = " << original << '\n';  // Vẫn là 5!

    // Gọi hàm có tham số không tên:
    logEvent(101, 999999);

    // =========================================================================
    // 7. CÁC BẪY BIÊN DỊCH CẦN TRÁNH (COMPILE ERRORS):
    // =========================================================================
    // add(5);       // LỖI: Too few arguments (đòi 2 mà chỉ đưa 1)
    // add(1, 2, 3); // LỖI: Too many arguments (đòi 2 mà đưa 3)

    return 0;
}
```

```cpp
//2.5
#include <iostream>

// Hai biến tham số 'x' và 'y' được tạo ra khi hàm add được gọi
int add(int x, int y) // 1. x tạo trước, 2. y tạo sau
{
    // Biến cục bộ 'result' được tạo tại điểm định nghĩa
    int result{ x + y }; // 3. result tạo sau cùng

    return result; 
} // THỨ TỰ HỦY BIẾN (LIFO - Vào sau ra trước):
  // result bị hủy đầu tiên -> đến y -> cuối cùng là x!

void printHello()
{
    // Đứng ở đây: biến 'a' của main() đang OUT OF SCOPE (không thể truy cập)
    // nhưng 'a' vẫn đang SỐNG trên RAM (chưa bị DESTROYED)!
    std::cout << "Hello from printHello()\n";
}

int main()
{
    // BEST PRACTICE: Khai báo biến càng gần nơi sử dụng đầu tiên càng tốt
    int a{ 5 }; 
    int b{ 10 };

    printHello(); // Trong suốt thời gian hàm này chạy, 'a' và 'b' vẫn sống trên RAM

    // FUNCTIONAL SEPARATION: 
    // Biến 'x' và 'y' ở main() dưới đây HOÀN TOÀN ĐỘC LẬP với 'x' và 'y' của hàm add()
    int x{ 100 };
    int y{ 200 };
    std::cout << "main x: " << x << ", main y: " << y << '\n';

    // TEMPORARY OBJECT (Đối tượng tạm thời):
    // add(a, b) trả về con số 15, lưu trong một đối tượng tạm không tên trên RAM.
    // Đối tượng tạm này bị HỦY ngay khi kết thúc dấu ';' của dòng lệnh này!
    std::cout << "Tong a + b: " << add(a, b) << '\n';

    return 0;
} // 'y', 'x', 'b', 'a' chính thức bị DESTROYED tại đây!

```

```cpp
//2.6
#include <iostream>

// =============================================================================
// KHỐI 1: INPUT (Đọc dữ liệu từ người dùng)
// Nhiệm vụ duy nhất: Hỏi và trả về số tiền hợp lệ
// =============================================================================
double getTransferAmount()
{
    std::cout << "Nhap so tien muon chuyen: ";
    double amount{};
    std::cin >> amount;
    return amount;
}

// =============================================================================
// KHỐI 2: PURE CALCULATION (Xử lý tính toán thuần túy)
// Không chứa std::cin hay std::cout -> Rất dễ viết Unit Test tự động!
// Có thể tái sử dụng cho Web, Mobile app, Console, hoặc lưu Database.
// =============================================================================
double calculateFee(double amount)
{
    return amount * 0.02; // Phí 2%
}

double calculateTotal(double amount, double fee)
{
    return amount + fee;
}

// =============================================================================
// KHỐI 3: OUTPUT (Hiển thị / Trình bày kết quả)
// Nhiệm vụ duy nhất: Định dạng dữ liệu và in ra màn hình
// =============================================================================
void printReceipt(double total, double fee)
{
    std::cout << "------------------------------\n";
    std::cout << "Tong tien phi phai tra la: " << fee << '\n';
    std::cout << "Tong tien phai tra la    : " << total << '\n';
    std::cout << "------------------------------\n";
}

// =============================================================================
// KHỐI 4: ORCHESTRATOR (Nhạc trưởng điều phối luồng chạy)
// Áp dụng tính Abstraction: main() chỉ gọi hàm, không cần biết chi tiết ruột hàm
// =============================================================================
int main()
{
    // Bước 1: Thu thập Input
    double amount{ getTransferAmount() };

    // Bước 2: Xử lý tính toán (Processing)
    double fee{ calculateFee(amount) };
    double total{ calculateTotal(amount, fee) };

    // Bước 3: Xuất kết quả (Output)
    printReceipt(total, fee);

    return 0;
}

```

```cpp
//2.7
#include <iostream>

// ==========================================================
// 1. FORWARD DECLARATIONS (FUNCTION PROTOTYPES)
// Lời hứa với Compiler: Khai báo tên, kiểu tham số, kiểu trả về
// ==========================================================
double processTransaction();
double calculateFee(double amount);
double calculation(double amount, double fee);
void printResult(double total, double fee);

// ==========================================================
// 2. MAIN FUNCTION (Bức tranh tổng quan của chương trình)
// ==========================================================
int main()
{
    double amount{ processTransaction() };
    double fee{ calculateFee(amount) };
    double total{ calculation(amount, fee) };

    printResult(total, fee);

    return 0;
}

// ==========================================================
// 3. FUNCTION DEFINITIONS (Chi tiết cách làm của từng hàm)
// ==========================================================
double processTransaction()
{
    std::cout << "Nhap so tien muon chuyen: ";
    double amount{};
    std::cin >> amount;
    return amount;
}

double calculateFee(double amount)
{
    return amount * 0.02;
}

double calculation(double amount, double fee)
{
    return amount + fee;
}

void printResult(double total, double fee)
{
    std::cout << "Tong tien phai tra la: " << total << '\n';
    std::cout << "Tong tien phi phai tra la: " << fee << '\n';
}
```

```cpp
//2.9 
#include <iostream>

// ============================================================================
// 1. NAMESPACE (KHÔNG GIAN TÊN) LÀ GÌ?
// Giúp tạo các vùng phạm vi riêng biệt để các định danh trùng tên không đụng độ nhau.
// ============================================================================
namespace Math
{
    // Hàm doSomething() nằm gọn trong vùng "Math"
    void doSomething()
    {
        std::cout << "[Math] Goi ham doSomething()\n";
    }
}

namespace Physics
{
    // Trùng tên "doSomething", nhưng KHÔNG hề xung đột vì nằm ở namespace khác!
    void doSomething()
    {
        std::cout << "[Physics] Goi ham doSomething()\n";
    }
}

// ============================================================================
// 2. GLOBAL NAMESPACE (KHÔNG GIAN TÊN TOÀN CỤC)
// Mọi hàm hoặc biến viết tự do bên ngoài đều thuộc về Global Namespace.
// ============================================================================
void doSomething()
{
    std::cout << "[Global] Goi ham doSomething() o Global Scope\n";
}

// ⚠️ QUY TẮC CỐT LÕI VỀ NỘI DUNG TRONG NAMESPACE:
// - Chỉ được chứa KHAI BÁO (Declarations) và ĐỊNH NGHĨA (Definitions).
// - Ví dụ: int x{ 5 };        --> HỢP LỆ (Định nghĩa biến).
// - Ví dụ: x = 10;            --> LỖI (Câu lệnh thực thi không được đứng ngoài hàm).
// - Ví dụ: std::cout << "Hi"; --> LỖI (Câu lệnh thực thi không được đứng ngoài hàm).

// ============================================================================
// 3. TOÁN TỬ PHÂN GIẢI PHẠM VI (SCOPE RESOLUTION OPERATOR: ::)
// ============================================================================
int main()
{
    // Cú pháp: [Tên_Namespace]::[Định_Danh]
    
    Math::doSomething();    // 1. Chỉ định rõ ràng: lấy hàm trong Math
    Physics::doSomething(); // 2. Chỉ định rõ ràng: lấy hàm trong Physics

    // Dấu :: đứng một mình ở đầu: ép buộc tìm trong Global Namespace
    ::doSomething();        // 3. Gọi hàm doSomething() ở phạm vi toàn cục

    // ========================================================================
    // 4. TẠI SAO BÀI TRỪ "using namespace std;"?
    // ========================================================================
    // - "cout", "cin", "endl" thực chất chỉ là tên gọi. Chúng sống trong namespace "std".
    // - Nếu viết "using namespace std;", bạn mở bung hàng nghìn công cụ của std ra.
    // - Khi đó, nếu bạn vô tình đặt tên biến/hàm là: count, min, max, size, data...
    //   Compiler sẽ lập tức quăng lỗi "reference is ambiguous" (lỗi mơ hồ).
    // - BEST PRACTICE: Luôn luôn viết tiền tố tường minh: std::cout, std::cin.

    std::cout << "\nChuong trinh chay hoan hao, khong he bi xung dot ten!\n";

    return 0;
}
```

```cpp
//2.13
// ============================================================================
// BÀI 2.13: HOW TO DESIGN YOUR FIRST PROGRAMS (THIẾT KẾ CHƯƠNG TRÌNH ĐẦU TIÊN)
// ============================================================================

// 1. TƯ DUY THIẾT KẾ PHÂN RÃ TỪ TRÊN XUỐNG (TOP-DOWN DECOMPOSITION)
// Bước 1: Xác định rõ Input, Output và các bước trung gian của bài toán.
// Bước 2: Viết hàm main() đóng vai trò Nhạc trưởng (Orchestrator).
//         main() chỉ điều phối các hàm con, đọc luồng chương trình rõ ràng.
// Bước 3: Hiện thực hóa từng hàm con (Bottom-up implementation).
//         Mỗi hàm/module chỉ giải quyết đúng 1 việc (Single Responsibility Principle).

// 2. PHÒNG THỦ VÀ KIỂM SOÁT LỖI (DEFENSIVE PROGRAMMING)
// - Luôn xác thực dữ liệu đầu vào (Input validation).
// - Kiểm tra các trường hợp biên nguy hiểm: chia cho 0.
// - Ép kiểu static_cast<double> để tránh phép chia số nguyên làm mất phần thập phân.
// - Sử dụng switch-case có break đầy đủ và nhánh default phòng thủ.

#include <iostream>

// --- MODULE 1: INPUT ---
int getInteger()
{
    std::cout << "Enter an integer: ";
    int x{};
    std::cin >> x;
    return x;
}

// --- MODULE 2: OPERATION & VALIDATION ---
char getOperation()
{
    char op{};
    std::cout << "Chon phep tinh (+, -, *, /): ";
    std::cin >> op;

    while (op != '+' && op != '-' && op != '*' && op != '/')
    {
        std::cout << "Phep tinh khong hop le, hay nhap lai: ";
        std::cin >> op;
    }
    return op;
}

// --- MODULE 3: CALCULATION & OUTPUT ---
void printResult(int a, char op, int b)
{
    switch (op)
    {
        case '+':
            std::cout << "Ket qua: " << a + b << '\n';
            break;
        case '-':
            std::cout << "Ket qua: " << a - b << '\n';
            break;
        case '*':
            std::cout << "Ket qua: " << a * b << '\n';
            break;
        case '/':
            if (b == 0)
            {
                std::cout << "Loi: Khong the chia cho 0!\n";
            }
            else
            {
                std::cout << "Ket qua: " << static_cast<double>(a) / b << '\n';
            }
            break;
        default:
            std::cout << "Loi: Phep tinh khong hop le!\n";
            break;
    }
}

// --- ORCHESTRATOR: MAIN ---
int main()
{
    int x{ getInteger() };
    char op{ getOperation() };
    int y{ getInteger() };

    printResult(x, op, y);
    return 0;
}
```

```cpp
//3.1
// ============================================================================
// BÀI 3.1: SYNTAX AND SEMANTIC ERRORS (LỖI CÚ PHÁP VÀ LỖI NGỮ NGHĨA)
// ============================================================================

// 1. SYNTAX ERRORS (LỖI CÚ PHÁP)
// - Xảy ra khi câu lệnh vi phạm văn phạm (Grammar) của ngôn ngữ C++.
// - Phát hiện ở khâu Lexer/Parser (trước khi kiểm tra ý nghĩa).
// - Compiler luôn phát hiện 100% và chặn không cho tạo file thực thi.
// - Ví dụ: thiếu dấu chấm phẩy (;), ngoặc không đóng mở đủ, biến bắt đầu bằng số (int 2x;).

// 2. COMPILE-TIME SEMANTIC ERRORS (LỖI NGỮ NGHĨA LÚC BIÊN DỊCH)
// - Cú pháp hoàn toàn đúng chuẩn grammar, nhưng sai quy tắc về mặt ý nghĩa ngôn ngữ:
//   + Không tương thích kiểu: int a { "hello" };
//   + Gán giá trị vào hằng số / rvalue: 5 = 10; (vế trái phải là modifiable lvalue).
//   + Sử dụng biến chưa khai báo.

// 3. RUNTIME SEMANTIC ERRORS (LỖI NGỮ NGHĨA LÚC CHẠY / LOGIC ERRORS)
// - Chương trình biên dịch thành công, nhưng khi chạy sinh ra hành vi không mong muốn:
//   + Crash: Phép chia cho 0 (Hardware trap SIGFPE), truy cập con trỏ rác.
//   + Sai kết quả (Silent Killer): Sai toán tử, sai độ ưu tiên phép toán.
//   + Mã chết (Dead Code): Lệnh return đặt trước các câu lệnh khác.
//   + Undefined Behavior (UB): Sử dụng biến chưa khởi tạo, tràn số nguyên có dấu.

// 4. TRIẾT LÝ ZERO-OVERHEAD CỦA C++ VÀ BẢO VỆ TRÀN SỐ (C++20 std::midpoint)
// - C++ không tự chèn lệnh if (b == 0) để giữ hiệu năng tối đa cho CPU.
// - Phép cộng (x + y) / 2 có thể bị tràn số nguyên (Signed Overflow - UB).
// - C++20 giới thiệu std::midpoint trong <numeric> giải quyết triệt để vấn đề này.

#include <iostream>
#include <numeric> // C++20: std::midpoint

// Hàm mô phỏng lỗi logic do độ ưu tiên toán tử:
int buggyAverage(int x, int y)
{
    return x + y / 2; // LỖI: Chia trước cộng sau -> (x + (y / 2))
}

// Hàm tính trung bình an toàn tuyệt đối, chống tràn số nguyên:
int safeAverage(int x, int y)
{
    return std::midpoint(x, y); // C++20: Khắc phục triệt để signed overflow
}

int main()
{
    std::cout << "=== BÀI 3.1: PHÂN BIỆT LỖI CÚ PHÁP & NGỮ NGHĨA ===\n\n";

    // 1. Minh họa lỗi logic âm thầm (Silent Semantic Error):
    int a{ 4 };
    int b{ 6 };
    std::cout << "1. Trung binh cong sai (buggyAverage): " << buggyAverage(a, b) 
              << " (Ky vong: 5, Thuc te: 7 do uu tien toan tu)\n";

    // 2. Minh họa xử lý an toàn chống tràn số lớn với std::midpoint (C++20):
    int large1{ 1'500'000'000 };
    int large2{ 1'800'000'000 };
    std::cout << "2. Trung binh 2 so cuc lon (std::midpoint C++20): " 
              << safeAverage(large1, large2) << '\n';

    std::cout << "\n-> Compiler bat 100% loi Syntax & Compile-time Semantic.\n"
              << "-> Lap trinh vien phai tu chu dong phong ngu Runtime Semantic Errors!\n";

    return 0;
}
```

```cpp
//3.2
// ============================================================================
// BÀI 3.2: THE DEBUGGING PROCESS (QUY TRÌNH GỠ LỖI CHUẨN MỰC)
// ============================================================================

// 1. PHÂN BIỆT SYMPTOM (TRIỆU CHỨNG) VS ROOT CAUSE (NGUYÊN NHÂN GỐC)
// - Symptom: Nơi lỗi phát tác bề ngoài (Crash, Exception, Sai số màn hình).
// - Root Cause: Dòng code đầu tiên làm sai luồng dữ liệu (Data flow).
// - NGUY HIỂM: Nơi xảy ra crash thường cách rất xa nguyên nhân gốc!
//   -> Tuyệt đối không "vá tạm" (Band-aid fix / Workaround) tại nơi crash.

// 2. QUY TRÌNH DEBUGGING 6 BƯỚC KHOA HỌC:
//   Bước 1: Find the root cause     - Lần ngược luồng dữ liệu để tìm nguyên nhân gốc.
//   Bước 2: Understand the problem  - Hiểu tường tận TẠI SAO nó lại xảy ra (tránh sửa mò).
//   Bước 3: Determine the fix       - Lên phương án sửa tận gốc, sạch và an toàn.
//   Bước 4: Repair the issue        - Tiến hành sửa code và biên dịch lại.
//   Bước 5: Retest the fix          - Chạy lại đúng ca lỗi ban đầu xem đã hết chưa.
//   Bước 6: Retest for regressions  - Kiểm thử hồi quy toàn bộ hệ thống (tránh lỗi mới).

// 3. THUẬT NGỮ CỐT TỬ: REGRESSION & REGRESSION TESTING
// - Regression (Lỗi hồi quy): Hiện tượng sửa lỗi A vô tình làm hỏng tính năng B.
// - Regression Testing: Chạy lại toàn bộ test case cũ để chứng minh code vẫn toàn vẹn.

#include <iostream>
#include <cassert>

// [KỊCH BẢN DEBUGGING]
// Giả sử ta có hàm tính chiết khấu.
// Code lỗi ban đầu (Root Cause): return basePrice - discountPercent; (sai công thức)
// Code đã sửa (Repair):
int calculateDiscountedPrice(int basePrice, int discountPercent)
{
    // Đúng công thức: giá sau giảm = gốc * (100 - %) / 100
    return basePrice * (100 - discountPercent) / 100;
}

// BƯỚC 5 & 6: BỘ TEST KIỂM THỬ XÁC NHẬN & KIỂM THỬ HỒI QUY
void runAllTests()
{
    // Bước 5: Retest the fix (Kiểm thử lại lỗi ban đầu bị báo cáo: món 100k giảm 20%)
    assert(calculateDiscountedPrice(100, 20) == 80);
    std::cout << "[Pass] Buoc 5: Retest the fix thanh cong (100 giam 20% = 80)\n";

    // Bước 6: Retest for regressions (Kiểm thử hồi quy các trường hợp biên cũ)
    assert(calculateDiscountedPrice(100, 0) == 100);  // Giảm 0%
    assert(calculateDiscountedPrice(100, 100) == 0);  // Giảm 100%
    assert(calculateDiscountedPrice(0, 50) == 0);     // Giá gốc 0
    std::cout << "[Pass] Buoc 6: Regression testing thanh cong! Khong sinh loi moi.\n";
}

int main()
{
    std::cout << "=== BÀI 3.2: DEMO QUY TRÌNH 6 BƯỚC DEBUGGING ===\n\n";
    runAllTests();
    std::cout << "\n-> Luon nho: Tim dung Root Cause, tranh sua vo Symptom!\n";
    return 0;
}
```

```cpp
//3.3
// ============================================================================
// BÀI 3.3: A STRATEGY FOR DEBUGGING (CHIẾN LƯỢC TRUY VẾT VÀ CÔ LẬP LỖI)
// ============================================================================

// 1. GIỚI HẠN CỦA VIỆC ĐỌC CODE BẰNG MẮT (CODE INSPECTION)
// - Chỉ áp dụng cho các hàm cực ngắn và đơn giản.
// - Bẫy "Giả định sai lầm" (Bad Assumptions): Khi viết code với giả định sai,
//   khi đọc lại ta sẽ lặp lại chính xác giả định đó -> mắt thường bị "mù" trước bug.

// 2. QUY TẮC VÀNG SỐ 1: TÁI HIỆN LỖI (REPRODUCE THE PROBLEM)
// - "If you can't reproduce the problem, you can't debug it!"
// - Reproduction Steps: Chuỗi thao tác, input, cấu hình cụ thể để lỗi lặp lại 100%.
//   Không đoán mò khi chưa nhìn thấy lỗi xảy ra trên máy của mình.

// 3. CHIẾN LƯỢC KHOANH VÙNG HI-LO (HOMING IN / BINARY SEARCH DEBUGGING)
// - Ranh giới nghi vấn: Từ đầu chương trình đến dòng xuất hiện triệu chứng đầu tiên.
// - Đặt trạm kiểm soát (Checkpoints/Logs) tại điểm giữa (Midpoint):
//   + Checkpoint còn đúng: "Too low"  -> Lỗi nằm ở nửa sau (loại bỏ nửa trước).
//   + Checkpoint bị sai : "Too high" -> Lỗi nằm ở nửa trước (loại bỏ nửa sau).
// - Hiệu năng: Cô lập 1 dòng code lỗi trong 1.000 dòng chỉ mất ~10 bước (log2(1000) ~ 10).
// - Liên hệ thực tế: Lệnh "git bisect" áp dụng 100% thuật toán Hi-Lo này.

#include <iostream>

// Mô phỏng 4 công đoạn trong dây chuyền dữ liệu:
int step1_LoadData()
{
    return 10; // Đọc dữ liệu ban đầu = 10 (Đúng)
}

int step2_Transform(int x)
{
    return x * 2; // Biến đổi: 10 * 2 = 20 (Đúng)
}

int step3_ApplyTax(int x)
{
    // [BUG TIỀM ẨN]: Trừ 50 thay vì cộng 5 -> Dữ liệu bị âm!
    return x - 50; 
}

int step4_Finalize(int x)
{
    return x + 100;
}

int main()
{
    std::cout << "=== BÀI 3.3: DEMO CHIẾN LƯỢC KHOANH VÙNG HI-LO (HOMING IN) ===\n\n";

    int data1{ step1_LoadData() };
    int data2{ step2_Transform(data1) };

    // --- [TRẠM KIỂM SOÁT HI-LO Ở ĐIỂM GIỮA (MIDPOINT)] ---
    // Kiểm tra trạng thái dữ liệu sau Step 2:
    std::cout << "[Hi-Lo Checkpoint] Sau Step 2, data2 = " << data2 << '\n';
    if (data2 == 20)
    {
        std::cout << "-> Ket qua van DUNG (Too low)! Loai bo Step 1 & Step 2 khoi dien nghi van.\n"
                  << "-> Thu hep pham vi dieu tra sang nua sau (Step 3 & Step 4)!\n\n";
    }

    int data3{ step3_ApplyTax(data2) }; // Bug xảy ra ở đây: 20 - 50 = -30
    int finalResult{ step4_Finalize(data3) };

    std::cout << "Ket qua cuoi cung: " << finalResult << " (Thuc te ky vong phai lon hon 120)\n";
    std::cout << "\n-> Nho Hi-Lo Checkpoint, ta tim ra Root Cause nam chinh xac tai Step 3!\n";

    return 0;
}
```

```cpp
//3.4
// ============================================================================
// BÀI 3.4: BASIC DEBUGGING TACTICS (CÁC CHIẾN THUẬT GỠ LỖI CƠ BẢN)
// ============================================================================

// 1. CHIẾN THUẬT 1: COMMENTING OUT CODE (VÔ HIỆU HÓA CODE TẠM THỜI)
// - Mục đích: Loại bỏ nghi phạm để thu hẹp phạm vi.
// - 3 tình huống:
//   + Lỗi biến mất: Đoạn code đó là "nghi phạm số 1" (chứa bug hoặc kích hoạt bug).
//   + Lỗi giữ nguyên: Đoạn code đó vô tội -> loại trừ.
//   + Lỗi biến hình sang lỗi mới (Dependency trap): Đoạn code đó làm nhiệm vụ thiết yếu
//     mà các hàm sau phụ thuộc vào. KHÔNG THỂ kết luận nó vô tội! Phải uncomment ngay!
// - Cảnh báo: Luôn kiểm tra "git diff" trước khi commit để tránh quên uncomment mã thật.

// 2. CHIẾN THUẬT 2: VALIDATING CODE FLOW (TRACING LUỒNG CHẠY)
// - In thông báo khi hàm bắt đầu chạy để biết hàm nào được gọi, gọi bao nhiêu lần.
// - std::cerr vs std::cout:
//   + std::cout có buffer (bộ đệm): Dữ liệu gom vào RAM, nếu chương trình crash ngay sau
//     lệnh in thì RAM bị giải phóng -> log không kịp in ra màn hình -> đánh lừa vị trí lỗi.
//   + std::cerr là unbuffered (không đệm): Đẩy thẳng ký tự ra màn hình ngay lập tức.
//     -> Luôn ưu tiên dùng std::cerr khi debug!
// - Cạm bẫy "Quên cặp ngoặc ()":
//   + Viết `getValue` thay vì `getValue()` không phải lỗi cú pháp mà là warning.
//   + Tên hàm suy giảm thành Function Pointer, tự động ép kiểu sang bool (true) -> in ra 1.
//   + Gây ảo tưởng là hàm đã chạy và trả về 1!

// 3. CHIẾN THUẬT 3: PRINTING VALUES (IN GIÁ TRỊ BIẾN VÀ TRẠNG THÁI)
// - In giá trị tham số và biến trung gian tại các checkpoint để phát hiện sai lệch dữ liệu.

// 4. BỐN NHƯỢC ĐIỂM LỚN CỦA PRINT DEBUGGING (LÝ DO CẦN DÙNG DEBUGGER CHUYÊN DỤNG)
// - Làm bẩn mã nguồn (code clutter).
// - Nguy cơ gây lỗi mới khi thêm/xóa code debug thủ công.
// - Tốn thời gian biên dịch lại (recompile overhead) mỗi lần muốn soi một biến mới.
// - Không tái sử dụng được (xóa đi là mất).

#include <iostream>

int add(int x, int y)
{
    // Chiến thuật 2 & 3: Tracing luồng chạy và in giá trị tham số với std::cerr (unbuffered)
    std::cerr << "[DEBUG FLOW] add() called with parameters: x = " << x << ", y = " << y << '\n';
    return x + y;
}

int calculateTotal(int itemPrice, int quantity)
{
    std::cerr << "[DEBUG FLOW] calculateTotal() called: itemPrice = " << itemPrice 
              << ", quantity = " << quantity << '\n';
    return itemPrice * quantity;
}

int main()
{
    std::cout << "=== BÀI 3.4: DEMO BASIC DEBUGGING TACTICS ===\n\n";

    int price{ 50 };
    int count{ 2 };

    // Minh họa Cạm bẫy quên ngoặc ():
    // Tên hàm add suy giảm thành pointer -> ép kiểu ngầm sang bool -> in ra 1
    std::cout << "Gia tri khi quen ngoac tron (add): " << (void*)add 
              << " -> khi in truc tiep se decay thanh bool (1)\n\n";

    // Trạm kiểm soát in biến (Printing values)
    int total{ calculateTotal(price, count) };
    std::cerr << "[DEBUG VALUE] main::total = " << total << '\n';

    int finalPrice{ add(total, 10) }; // Cộng thêm phí ship 10
    std::cerr << "[DEBUG VALUE] main::finalPrice = " << finalPrice << '\n';

    std::cout << "\nTong chi phi cuoi cung: " << finalPrice << '\n';
    std::cout << "-> Toan bo debug tren duoc thuc hien bang std::cerr (Unbuffered)!\n";

    return 0;
}
```

```cpp
//3.6
// ============================================================================
// BÀI 3.6: USING AN INTEGRATED DEBUGGER: STEPPING (ĐI TỪNG BƯỚC)
// ============================================================================

// 1. TRẠNG THÁI CHƯƠNG TRÌNH (PROGRAM STATE)
// - Tại mỗi thời điểm chạy, chương trình lưu trữ:
//   + Giá trị của các biến trong bộ nhớ.
//   + Call stack (hàm nào đang gọi hàm nào).
//   + Điểm thực thi hiện tại (Current point of execution).
// - Debugger giúp ta: Đóng băng chương trình, kiểm soát luồng chạy và xem/sửa State.

// 2. MŨI TÊN THỰC THI (EXECUTION MARKER / YELLOW ARROW)
// - Luôn trỏ vào câu lệnh SẮP SỬA ĐƯỢC CHẠY TIẾP THEO (chưa chạy!).
// - Biến khai báo ở dòng đó chưa được gán giá trị mới.

// 3. BỘ BA LỆNH STEPPING (ĐI TỪNG BƯỚC):
// - Step Into (F11 / Shift+F7):
//   + Chạy 1 câu lệnh. Gặp hàm thì NHẢY VÀO TRONG thân hàm đó.
//   + Cẩn thận: Có thể nhảy vào code thư viện chuẩn (như operator<<).
// - Step Over (F10 / F7):
//   + Chạy 1 câu lệnh. Gặp hàm thì CHẠY NGẦM HẾT HÀM ĐÓ và dừng ở dòng lệnh tiếp theo.
//   + Dùng khi: Hàm đã đúng, không cần soi chi tiết.
// - Step Out (Shift+F11 / Ctrl+F7):
//   + Chạy nốt toàn bộ phần còn lại của hàm hiện tại và trở về hàm gọi nó.
//   + Dùng khi: Lỡ Step Into vào nhầm hàm hoặc muốn thoát ra nhanh.

// 4. VÌ SAO TUYỆT ĐỐI KHÔNG DEBUG TRÊN RELEASE BUILD?
// - Cờ tối ưu hóa (-O2 / -O3) làm thay đổi hoàn toàn mã máy:
//   + Xóa biến (Eliminate unused/constant variables -> báo <optimized out>).
//   + Nội tuyến hóa (Inlining -> hàm bị gộp vào chỗ gọi, không thể Step Into).
//   + Đảo thứ tự lệnh (Instruction Reordering -> mũi tên nhảy loạn xạ).
// - Thiếu Debug Symbols (cờ -g): Cắt bỏ thông tin ánh xạ mã nguồn .cpp với mã máy.
// -> Luôn build ở chế độ Debug: `g++ -g -O0 file.cpp -o file`

#include <iostream>

void printDetails(int id, double balance)
{
    // Dòng 1 bên trong printDetails:
    // Nếu từ main() bấm "Step Into" -> Mũi tên dừng ở đây!
    // Nếu muốn thoát nhanh về main() -> Bấm "Step Out"!
    std::cout << "User ID: " << id << " | Balance: $" << balance << '\n';
}

double calculateBonus(double salary)
{
    double bonusRate{ 0.15 };
    return salary * bonusRate;
}

int main()
{
    std::cout << "=== BÀI 3.6: DEMO INTEGRATED DEBUGGER - STEPPING ===\n\n";

    int userId{ 101 };
    double baseSalary{ 5000.0 };

    // TÌNH HUỐNG STEPPING MINH HỌA:
    // Giả sử mũi tên đang dừng ở dòng dưới:
    // 1. Nếu bấm Step Over (F10): Hàm calculateBonus chạy ngầm, 
    //    tính ra bonus = 750 và mũi tên nhảy thẳng xuống lệnh printDetails tiếp theo.
    // 2. Nếu bấm Step Into (F11): Mũi tên nhảy vào dòng "double bonusRate{ 0.15 };" bên trong calculateBonus.
    double bonus{ calculateBonus(baseSalary) };

    double totalEarnings{ baseSalary + bonus };

    // Gọi hàm in thông tin:
    printDetails(userId, totalEarnings);

    std::cout << "\nChuong trinh da hoan tat sau cac buoc Stepping!\n";
    return 0;
}
```

```cpp
//3.7
// ============================================================================
// BÀI 3.7: USING AN INTEGRATED DEBUGGER: RUNNING & BREAKPOINTS
// ============================================================================

// 1. RUN TO CURSOR (CHẠY TỚI CON TRỎ CHUỘT)
// - Mục đích: Cho chương trình chạy tốc độ tối đa đến vị trí con trỏ chuột rồi dừng.
// - Tính chất: CHỈ DỪNG ĐÚNG 1 LẦN (One-time).
//   + Khi bấm Continue (F5), chương trình chạy tiếp đến hết mà KHÔNG dừng lại ở dòng đó
//     nữa, dù dòng đó có lặp lại nhiều lần.

// 2. BREAKPOINTS (ĐIỂM DỪNG CỐ ĐỊNH - PHÍM F9)
// - Mục đích: Đặt chốt chặn cố định tại một dòng lệnh.
// - Tính chất: BỀN VỮNG (Persistent) & DỪNG MỌI LẦN CHẠY QUA (Hits every time).
//   + Nếu đặt trong vòng lặp 10 lần -> Bấm Continue (F5) sẽ dừng lại đủ cả 10 lần.
// - Ứng dụng phát hiện "Mã chết" (Dead Code):
//   + Breakpoint chỉ kích hoạt khi CPU thực sự bước chân tới.
//   + Đặt breakpoint trong nhánh if mà không thấy dừng -> biết chắc 100% nhánh đó không chạy!

// 3. JUMP TO CURSOR / SET NEXT STATEMENT (NHẢY CÓC CON TRỎ LỆNH)
// - Thay đổi điểm thực thi tiến hoặc lùi mà không chạy các lệnh ở giữa.
// - CẢNH BÁO TỐI QUAN TRỌNG:
//   + Lệnh này CHỈ dời con trỏ lệnh, KHÔNG HOÀN TÁC TRẠNG THÁI BỘ NHỚ RAM!
//   + Ví dụ: x = x + 10 (chạy xong x = 10). Nếu Jump lùi lại chạy lại -> x = 20!
//   + Khác với Step Back (Reverse Debugging): Step Back hoàn tác cả RAM (x quay về 0).

#include <iostream>

void handleOrder(int orderId)
{
    // [THỰC HÀNH BREAKPOINT]
    // Đặt breakpoint tại đây (F9):
    // Vì main() gọi 3 lần, mỗi lần bấm Continue (F5) debugger sẽ dừng lại ở đây!
    std::cout << "Dang xu ly don hang #" << orderId << '\n';
}

int main()
{
    std::cout << "=== BÀI 3.7: DEMO RUNNING & BREAKPOINTS ===\n\n";

    // 1. Minh họa Breakpoint dừng nhiều lần:
    for (int i{ 1 }; i <= 3; ++i)
    {
        handleOrder(i);
    }

    // 2. Minh họa Unreachable Code:
    if (1 > 100)
    {
        // Đặt Breakpoint ở đây: Debugger KHÔNG BAO GIỜ DỪNG vì CPU không nhảy vào!
        std::cout << "Dieu nay khong bao gio xay ra!\n";
    }

    // 3. Minh họa Jump to Cursor & Biến thiên bộ nhớ:
    int counter{ 0 };
    counter = counter + 10; // Chạy xong counter = 10.
    // Nếu dùng Jump to Cursor quay lại dòng trên và chạy tiếp: counter sẽ thành 20!

    std::cout << "\nGia tri counter hien tai: " << counter << '\n';
    std::cout << "Hoan tat bai hoc ve Breakpoints & Control Flow!\n";

    return 0;
}
```

```cpp
//3.8
// ============================================================================
// BÀI 3.8: USING AN INTEGRATED DEBUGGER: WATCHING VARIABLES
// ============================================================================

// 1. CÁC CÁCH QUAN SÁT BIẾN TRONG DEBUGGER:
// - Mouse Hover: Rê chuột xem nhanh, nhưng phải rê lại mỗi bước.
// - Locals / Variables Window: Tự động liệt kê tất cả biến trong scope hiện tại.
// - Watch Window: Nhập biến/biểu thức cần theo dõi liên tục.
//   + Khi biến ra khỏi Scope (hàm kết thúc): chuyển sang <not available>,
//     khi vào lại scope sẽ tự động hiện lại giá trị!

// 2. ĐÁNH GIÁ BIỂU THỨC TRONG WATCH (EVALUATING EXPRESSIONS):
// - Có thể gõ: x + 10, x > 50, a * b -> Debugger tự tính toán kết quả tức thì.
// - CẢNH BÁO LIVE STATE MODIFICATION (CAN THIỆP TRỰC TIẾP VÀO RAM):
//   + Nếu gõ biểu thức gán: x = 100 trong Watch hoặc Debug Console,
//     Debugger THỰC SỰ GHI ĐÈ GIÁ TRỊ 100 VÀO RAM CỦA BIẾN x!
//   + Lợi ích: Thử nghiệm kịch bản nóng mà không cần sửa code / biên dịch lại.
//   + Nguy cơ: Gõ nhầm phép so sánh (==) thành phép gán (=) sẽ làm sai lệch dữ liệu thật.

// 3. DATA BREAKPOINT / WATCHPOINT ("BREAK WHEN VALUE CHANGES"):
// - Dừng chương trình ngay tại khoảnh khắc ô nhớ của biến bị ghi đè/thay đổi.
// - Sử dụng cơ chế phần cứng của CPU (Hardware watchpoint).
// - Cực kỳ đắc lực để bắt "kẻ giấu mặt" ghi đè bậy vào biến trong dự án lớn.

#include <iostream>

void modifyInventory(int& stock, int soldCount)
{
    // [THỰC HÀNH WATCH TRONG HÀM]
    // Biến stock tham chiếu đến currentStock ở main.
    // Nếu đặt Data Breakpoint trên biến currentStock:
    // Ngay khoảnh khắc dòng dưới chạy -> Debugger lập tức dừng lại!
    stock -= soldCount;
}

int main()
{
    std::cout << "=== BÀI 3.8: DEMO WATCHING VARIABLES ===\n\n";

    int currentStock{ 100 };
    int bonusGift{ 10 };

    // [THỬ NGHIỆM TRONG WATCH WINDOW]:
    // 1. Thêm 'currentStock' -> Thấy giá trị 100.
    // 2. Thêm biểu thức 'currentStock + bonusGift' -> Debugger tính ra 110.
    // 3. Thêm biểu thức điều kiện: 'currentStock < 50' -> Kết quả: false.
    // 4. [CAN THIỆP RAM]: Thử gõ trong Debug Console: 'currentStock = 200'
    //    -> Giá trị biến trong RAM đổi ngay lập tức thành 200!

    std::cout << "So luong ton kho ban dau: " << currentStock << '\n';

    modifyInventory(currentStock, 30); // stock giam con 70 (Watch boi do)

    std::cout << "So luong sau khi ban: " << currentStock << '\n';

    return 0;
}
```

```cpp
//3.9
// ============================================================================
// BÀI 3.9: USING AN INTEGRATED DEBUGGER: THE CALL STACK
// ============================================================================

// 1. CALL STACK (NGĂN XẾP CUỘC GỌI) LÀ GÌ?
// - Là danh sách ghi nhận các hàm đang hoạt động (active functions) theo cơ chế LIFO.
// - Đỉnh ngăn xếp (Top): Là hàm đang được CPU thực thi trực tiếp tại thời điểm hiện tại.
// - Khi hàm con gặp lệnh return -> Bị POP khỏi Call Stack, CPU quay về hàm cha ngay bên dưới.

// 2. MŨI TÊN VÀNG VS MŨI TÊN XANH TRONG VS CODE:
// - Mũi tên vàng (Execution marker): Chỉ dòng lệnh sắp chạy ở hàm hiện tại (đỉnh Call Stack).
// - Mũi tên xanh (Call Frame marker): Xuất hiện khi bạn nhấp vào một hàm cha trong Call Stack.
//   -> Mũi tên xanh chỉ ĐIỂM GỌI (Call Site) và cũng là ĐIỂM HẸN QUAY VỀ của hàm cha đó.

// 3. CHUYỂN ĐỔI NGỮ CẢNH STACK FRAME (CONTEXT SWITCHING):
// - Mỗi hàm trên Call Stack có một không gian bộ nhớ riêng (Stack Frame).
// - Click vào 'c' -> Cửa sổ Locals hiện biến của 'c'.
// - Click vào 'main' -> Cửa sổ Locals chuyển sang hiện biến của 'main'.
// - Cực kỳ hữu dụng để kiểm tra giá trị tham số mà hàm cha đã truyền vào hàm con!

#include <iostream>

void c(int paramC)
{
    // [ĐỈNH STACK] Mũi tên vàng dừng ở đây nếu đặt breakpoint
    int localVarC{ paramC + 100 };
    std::cout << "Gia tri tai c(): " << localVarC << '\n';
}

void b(int paramB)
{
    int localVarB{ paramB * 2 };
    c(localVarB); // [CALL SITE] Khi click 'b' tren Call Stack, mui ten xanh tro o day!
}

void a(int paramA)
{
    int localVarA{ paramA + 5 };
    b(localVarA); // [CALL SITE] Khi click 'a' tren Call Stack, mui ten xanh tro o day!
}

int main()
{
    std::cout << "=== BÀI 3.9: DEMO THE CALL STACK ===\n\n";

    int rootData{ 42 };

    // [THỰC HÀNH CALL STACK]:
    // 1. Đặt breakpoint tại hàm c(), bấm F5.
    // 2. Thứ tự Call Stack: c() -> b() -> a() -> main().
    // 3. Click vào 'main': Editor hiện mũi tên xanh tại dòng 'a(rootData);'
    //    và cửa sổ Locals lập tức hiển thị 'rootData = 42'!
    a(rootData);

    std::cout << "Hoan thanh chuoi goi ham thanh cong!\n";
    return 0;
}
```

```cpp
//3.10
// ============================================================================
// BÀI 3.10: FINDING ISSUES BEFORE THEY BECOME PROBLEMS
// ============================================================================

// 1. NGUYÊN TẮC HAI CHIẾC MŨ (THE TWO HATS RULE):
// - Mũ 1: Thêm tính năng mới (Behavioral change).
// - Mũ 2: Tái cấu trúc mã (Structural change / Refactoring).
// -> TUYỆT ĐỐI KHÔNG đội 2 mũ cùng lúc! Hoặc Refactor trước -> Test -> Thêm tính năng,
//    hoặc Thêm tính năng -> Test -> Refactor. Nếu làm đồng thời, khi lỗi xuất hiện
//    bạn sẽ không biết do logic tính năng mới hay do chia tách hàm bị sai.

// 2. KIỂM THỬ ĐƠN VỊ TỰ ĐỘNG (UNIT TESTING):
// - Viết hàm kiểm tra riêng biệt cho từng hàm độc lập.
// - Chạy trong 1/1000 giây, tự động phát hiện lỗi hồi quy (Regression) mỗi khi sửa code.

// 3. LẬP TRÌNH PHÒNG THỦ (DEFENSIVE PROGRAMMING):
// - Đón đầu lỗi do truyền đối số sai, kiểm tra điều kiện đầu vào (Preconditions).

// 4. PHÂN TÍCH TĨNH (STATIC ANALYSIS):
// - Compiler (-Wall -Wextra) và Linter (clang-tidy) quét mã nguồn mà không cần chạy chương trình:
//   + Bắt biến chưa khởi tạo (Data flow analysis)
//   + Bắt nhánh code chết không thể chạy tới (Control flow analysis)

#include <iostream>

// Hàm được thiết kế nhỏ gọn (< 10 dòng)
int divide(int numerator, int denominator)
{
    // Lập trình phòng thủ: Kiểm tra mẫu số trước khi tính toán
    if (denominator == 0)
    {
        std::cerr << "Loi phong thu: Khong the chia cho 0!\n";
        return 0; // Xử lý an toàn tạm thời
    }

    return numerator / denominator;
}

// Hàm Unit Test tự động cho hàm divide()
void testDivide()
{
    std::cout << "--- CHAY UNIT TEST CHO divide() ---\n";
    std::cout << "Test 10 / 2 == 5: " << (divide(10, 2) == 5 ? "PASS" : "FAIL") << '\n';
    std::cout << "Test -6 / 3 == -2: " << (divide(-6, 3) == -2 ? "PASS" : "FAIL") << '\n';
    std::cout << "Test 0 / 5 == 0: " << (divide(0, 5) == 0 ? "PASS" : "FAIL") << '\n';
    std::cout << "Test 8 / 0 (phong thu): " << (divide(8, 0) == 0 ? "PASS" : "FAIL") << '\n';
}

int main()
{
    std::cout << "=== BÀI 3.10: FINDING ISSUES BEFORE THEY BECOME PROBLEMS ===\n\n";

    // Thay vì gõ tay cin, ta kích hoạt bộ unit test tự động:
    testDivide();

    std::cout << "\nTat ca cac bai kiem tra da hoan tat nhanh chong va an toan!\n";
    return 0;
}
```

```cpp
//4.1
// ============================================================================
// BÀI 4.1: INTRODUCTION TO FUNDAMENTAL DATA TYPES
// ============================================================================

// 1. TẦNG 1: BẢN CHẤT BỘ NHỚ VÀ CÁC KIỂU FUNDAMENTAL
// - Máy tính đánh địa chỉ theo từng Byte (1 byte = 8 bits trên kiến trúc hiện đại).
// - Data Type cung cấp 3 thông tin cốt lõi cho compiler:
//   + Kích thước (Size): Cần cấp phát bao nhiêu byte RAM.
//   + Mã hóa & Phiên dịch (Interpretation): Đọc bit thành số bù 2, số thực IEEE 754 hay ký tự ASCII.
//   + Tập phép toán hợp lệ (Operations).
//
// 2. PHÂN BIỆT INTEGER TYPES VS INTEGRAL TYPES:
// - Standard Integer: short, int, long, long long.
// - Integral Types: Standard Integer + char + bool.
//   -> Tất cả đều lưu dưới RAM như số nguyên, nhưng compiler gán ngữ nghĩa hiển thị khác nhau.
//
// 3. TẦNG 2: GÓC NHÌN ADVANCED (ISO C++ STANDARD)
// - Object Representation vs Value Representation:
//   + bool chỉ cần 1 bit để chứa 0 hoặc 1, nhưng chiếm trọn 1 byte (8 bits). 7 bit còn lại là padding.
// - Incomplete Type: void đại diện cho sự vắng mặt của kiểu/giá trị, không thể tạo biến void.
// - Undefined Behavior (UB): Vi phạm biểu diễn giá trị hợp lệ (như nhét byte 2 vào bool)
//   sẽ khiến trình biên dịch tối ưu hóa sai lệch logic (-O0 vs -O2).

#include <iostream>
#include <cstring>

void demoFundamentalTypes()
{
    std::cout << "--- 1. DEMO KIỂU INTEGRAL (CÙNG LÀ SỐ NGUYÊN DƯỚI RAM) ---\n";
    int intVal{ 65 };
    char charVal{ 65 }; // Bảng mã ASCII: 65 là 'A'
    bool boolVal{ true }; // Dưới RAM lưu là 1

    std::cout << "int 65:    " << intVal << " (Kich thuoc: " << sizeof(intVal) << " bytes)\n";
    std::cout << "char 65:   " << charVal << " (Kich thuoc: " << sizeof(charVal) << " byte)\n";
    std::cout << "bool true: " << boolVal << " (Kich thuoc: " << sizeof(boolVal) << " byte)\n\n";

    std::cout << "--- 2. DEMO FLOATING-POINT TYPES ---\n";
    float fVal{ 3.14159f };
    double dVal{ 3.141592653589793 };
    std::cout << "float:  " << fVal << " (" << sizeof(fVal) << " bytes)\n";
    std::cout << "double: " << dVal << " (" << sizeof(dVal) << " bytes)\n\n";

    std::cout << "--- 3. DEMO ADVANCED: OBJECT REPRESENTATION & UNDEFINED BEHAVIOR ---\n";
    unsigned char fakeByte{ 2 }; // Giá trị bất hợp pháp cho bool
    bool hackedBool;
    std::memcpy(&hackedBool, &fakeByte, sizeof(hackedBool));

    std::cout << "Gia tri hackedBool in ra: " << hackedBool << '\n';
    std::cout << "Kiem tra hackedBool == true:  " << (hackedBool == true ? "TRUE" : "FALSE") << '\n';
    std::cout << "Kiem tra hackedBool == false: " << (hackedBool == false ? "TRUE" : "FALSE") << '\n';
    std::cout << "=> CHÚ Ý: Ket qua kiem tra se bi dao lon giua co -O0 va -O2 do Undefined Behavior!\n";
}

int main()
{
    std::cout << "=== BÀI 4.1: INTRODUCTION TO FUNDAMENTAL DATA TYPES ===\n\n";
    demoFundamentalTypes();
    return 0;
}
```

```cpp
//4.2
// ============================================================================
// BÀI 4.2: VOID
// ============================================================================

// 1. TẦNG 1: Ý NGHĨA VÀ 2 NGỮ CẢNH PHỔ BIẾN
// - void mang ý nghĩa là "no type" (không có kiểu gì cả / vô kiểu).
// - Incomplete Type: Compiler biết void tồn tại nhưng không thể biết kích thước để cấp phát RAM.
//   -> Không thể tạo biến void x; (Báo lỗi: variable has incomplete type 'void').
// - Ngữ cảnh 1: Hàm không trả về giá trị (void functions)
//   -> Không được return giá trị (return 5; là lỗi). Có thể dùng 'return;' rỗng để thoát sớm.
// - Ngữ cảnh 2: Hàm không nhận tham số
//   -> Trong C cũ: int foo(void).
//   -> Trong C++ hiện đại: int foo() với ngoặc rỗng là chuẩn nhất; dùng (void) bị coi là deprecated.

// 2. TẦNG 2: GÓC NHÌN ADVANCED (ISO C++ STANDARD)
// - Incomplete Types trong C++:
//   + Tạm thời không hoàn chỉnh (Temporarily incomplete): Forward declaration class Foo; (hoàn chỉnh sau).
//   + Vĩnh viễn không hoàn chỉnh (Permanently incomplete): Duy nhất chỉ có void (không bao giờ có kích thước).
// - Con trỏ void (void* ptr):
//   + void x; bị cấm vì không xác định kích thước.
//   + void* ptr; LẠI HỢP LỆ vì ptr là con trỏ lưu địa chỉ RAM (luôn là 8 bytes trên máy 64-bit).
//   + Không thể giải tham chiếu (*ptr) trực tiếp vì compiler không biết kiểu dữ liệu tại đó để dịch bit.

#include <iostream>

// Hàm void: Không trả về giá trị
void printMessage(bool shouldPrint)
{
    if (!shouldPrint)
    {
        std::cout << "[printMessage] Thoat som khoi ham bang 'return;'\n";
        return; // Thoát sớm hợp lệ
    }

    std::cout << "[printMessage] Xin chao tu ham void!\n";
}

// Best practice C++: Cặp ngoặc rỗng đại diện cho không nhận tham số
int getMagicNumber()
{
    return 42;
}

void demoVoidConcepts()
{
    std::cout << "--- 1. DEMO VOID FUNCTION ---\n";
    printMessage(false);
    printMessage(true);

    std::cout << "\n--- 2. DEMO BEST PRACTICE KHÔNG DÙNG (void) ---\n";
    std::cout << "Gia tri tu getMagicNumber(): " << getMagicNumber() << "\n\n";

    std::cout << "--- 3. DEMO INCOMPLETE TYPE VS VOID POINTER ---\n";
    // void x; // LỖI BIÊN DỊCH: variable has incomplete type 'void'

    void* genericPtr{ nullptr }; // HỢP LỆ vì con trỏ chỉ lưu địa chỉ ô nhớ
    std::cout << "Kich thuoc con tro void* tren he thong 64-bit: " << sizeof(genericPtr) << " bytes\n";
    std::cout << "=> Compiler biet kich thuoc dia chi (8 bytes) du kieu dich den la void!\n";
}

int main()
{
    std::cout << "=== BÀI 4.2: VOID ===\n\n";
    demoVoidConcepts();
    return 0;
}
```

```cpp
//4.3
// ============================================================================
// BÀI 4.3: OBJECT SIZES AND THE SIZEOF OPERATOR
// ============================================================================

// 1. TẦNG 1: SỐ LƯỢNG GIÁ TRỊ & TOÁN TỬ SIZEOF
// - n bits lưu trữ được 2^n giá trị riêng biệt:
//   + 1 byte (8 bits) = 256 giá trị.
//   + 2 bytes (16 bits) = 65,536 giá trị.
//   + 4 bytes (32 bits) = ~4.29 tỷ giá trị.
// - Toán tử sizeof: Đo kích thước kiểu hoặc biến theo đơn vị byte (sizeof(int), sizeof(x)).

// 2. TẦNG 2: GÓC NHÌN ADVANCED (ISO C++ STANDARD & PHẦN CỨNG)
// - Kích thước tối thiểu (Minimum Size) vs Kích thước thông thường:
//   + Chuẩn C++ chỉ quy định int >= 16 bits (2 bytes). Trên vi điều khiển (Arduino), int là 2 bytes!
//   + Quy tắc bất biến: 1 == sizeof(char) <= sizeof(short) <= sizeof(int) <= sizeof(long) <= sizeof(long long).
// - Cạm bẫy mô hình dữ liệu:
//   + Linux 64-bit (LP64): sizeof(long) = 8 bytes.
//   + Windows 64-bit (LLP64): sizeof(long) = 4 bytes!
// - Bản chất Compile-Time & Unevaluated Operand của sizeof:
//   + sizeof được tính toán ở compile-time (zero runtime overhead).
//   + Biểu thức bên trong sizeof KHÔNG BAO GIỜ chạy (unevaluated): sizeof(a++) không làm tăng a.
// - Hiệu năng CPU: Kiểu nhỏ hơn (short, char) chưa chắc nhanh hơn int do căn chỉnh thanh ghi CPU.

#include <iostream>
#include <climits> // CHAR_BIT

void demoObjectSizesAndSizeof()
{
    std::cout << "--- 1. ĐO KÍCH THƯỚC CÁC KIỂU DỮ LIỆU CƠ BẢN ---\n";
    std::cout << "1 Byte tren may nay co: " << CHAR_BIT << " bits\n";
    std::cout << "bool:        " << sizeof(bool) << " byte\n";
    std::cout << "char:        " << sizeof(char) << " byte (chuan C++ quy dinh luon = 1)\n";
    std::cout << "short:       " << sizeof(short) << " bytes\n";
    std::cout << "int:         " << sizeof(int) << " bytes (chuan ISO chi bat buoc >= 2 bytes)\n";
    std::cout << "long:        " << sizeof(long) << " bytes (8 tren Linux LP64, 4 tren Windows LLP64)\n";
    std::cout << "long long:   " << sizeof(long long) << " bytes (luon >= 8 bytes)\n";
    std::cout << "float:       " << sizeof(float) << " bytes\n";
    std::cout << "double:      " << sizeof(double) << " bytes\n\n";

    std::cout << "--- 2. DEMO TÍNH CHẤT UNEVALUATED OPERAND CỦA sizeof ---\n";
    int counter{ 5 };
    std::cout << "Gia tri counter ban dau: " << counter << '\n';
    std::cout << "sizeof(counter++):       " << sizeof(counter++) << " bytes\n";
    std::cout << "Gia tri counter sau do:  " << counter << " (VAN LA 5, counter++ KHONG he duoc chay!)\n";
}

int main()
{
    std::cout << "=== BÀI 4.3: OBJECT SIZES AND THE SIZEOF OPERATOR ===\n\n";
    demoObjectSizesAndSizeof();
    return 0;
}
```

```cpp
//4.4
// ============================================================================
// BÀI 4.4: SIGNED INTEGERS
// ============================================================================

// 1. TẦNG 1: KIỂU DỮ LIỆU CƠ BẢN & PHÉP CHIA NGUYÊN (BEGINNER)
// - 4 kiểu số nguyên có dấu: short (>=16b), int (>=16b, thường 32b), long (>=32b), long long (>=64b).
// - Quy ước đặt tên (Naming Conventions): Bỏ 'signed' và 'int' thừa: viết short, int, long, long long.
// - Dải giá trị n-bit: [-(2^(n-1)), 2^(n-1) - 1]. Cận dương ít hơn 1 do số 0 chiếm 1 slot ở phía không-âm.
// - Integer Division: Cắt bỏ phần thập phân về 0 (truncate towards zero), KHÔNG phải làm tròn xuống (floor)!
//   + 8 / 5 = 1; -8 / 5 = -1.

// 2. TẦNG 2: GÓC NHÌN CHUYÊN SÂU (ADVANCED & C++20 STANDARD)
// - Two's Complement (Số bù 2): Bắt buộc trong C++20 (P0907R4).
//   + Đổi dấu: Đảo bit + 1 (~x + 1). Chỉ có duy nhất 1 số 0.
//   + Tận dụng trạng thái 1000... cho số âm cực tiểu (-2^(n-1)).
//   + Phần cứng: A - B = A + (~B + 1), dùng chung 1 mạch cộng ALU (Adder).
// - Signed Integer Overflow là UNDEFINED BEHAVIOR (UB):
//   + Compiler (-O2) giả định UB không bao giờ xảy ra: (x + 1 > x) luôn được tối ưu thành 'true' kể cả khi x là INT_MAX!
//   + Nguy cơ: Code chạy ra số âm khi Debug (-O0) nhưng sai logic khi Release (-O2) do compiler loại bỏ code kiểm tra.

#include <iostream>
#include <climits>

bool checkOverflowAssumption(int x)
{
    // Compiler tối ưu (-O2) giả định UB không bao giờ xảy ra
    // => Biểu thức (x + 1 > x) được rút gọn thành 'true' ngay ở compile-time
    return (x + 1 > x);
}

void demoSignedIntegers()
{
    std::cout << "--- 1. CÁC KIỂU SỐ NGUYÊN CÓ DẤU & DẢI GIÁ TRỊ ---\n";
    std::cout << "short:     [" << SHRT_MIN << " den " << SHRT_MAX << "]\n";
    std::cout << "int:       [" << INT_MIN << " den " << INT_MAX << "]\n";
    std::cout << "long long: [" << LLONG_MIN << " den " << LLONG_MAX << "]\n\n";

    std::cout << "--- 2. PHÉP CHIA SỐ NGUYÊN (TRUNCATION TOWARDS ZERO) ---\n";
    std::cout << " 8 / 5  = " << (8 / 5) << "  (1.6 bi cat bo .6 huong ve 0)\n";
    std::cout << "-8 / 5  = " << (-8 / 5) << " (-1.6 bi cat bo .6 huong ve 0, KHONG phai floor ra -2)\n\n";

    std::cout << "--- 3. SIGNED OVERFLOW LÀ UNDEFINED BEHAVIOR (UB) ---\n";
    int maxInt{ 2'147'483'647 };
    std::cout << "maxInt: " << maxInt << '\n';
    std::cout << std::boolalpha;
    std::cout << "Gia tri checkOverflowAssumption(maxInt): " 
              << checkOverflowAssumption(maxInt) 
              << " (Compiler toi uu xem nhu luon true vi gia dinh khong co UB!)\n";
}

int main()
{
    std::cout << "=== BÀI 4.4: SIGNED INTEGERS ===\n\n";
    demoSignedIntegers();
    return 0;
}
```

```cpp
//4.5
// ============================================================================
// BÀI 4.5: UNSIGNED INTEGERS, AND WHY TO AVOID THEM
// ============================================================================

// 1. TẦNG 1: SỐ NGUYÊN KHÔNG DẤU & QUAY VÒNG MODULO (BEGINNER)
// - Dải giá trị n-bit: [0, 2^n - 1] (gấp đôi cận dương so với signed cùng size).
// - Cơ chế Modulo Wrap-around (Chuẩn hóa ISO, KHÔNG phải UB):
//   + Vượt cận trên: 65535 + 1 -> 0 (unsigned short).
//   + Vượt cận dưới: 0 - 1 -> 4294967295 (unsigned int 32-bit).

// 2. TẦNG 2: CẠM BẪY HỆ THỐNG & TRIẾT LÝ THIẾT KẾ (ADVANCED)
// - Cạm bẫy 1: Cận dưới là 0 rất gần với giá trị thực tế -> Dễ underflow (ví dụ: 2u - 3u = 4294967295).
// - Cạm bẫy 2: Ép kiểu ngầm định khi so sánh Signed vs Unsigned (Usual arithmetic conversions):
//   + (-1 < 1u) trả về FALSE vì -1 bị ép sang unsigned thành 4294967295!
// - Cạm bẫy 3: Vòng lặp vô tận khi đếm lùi: (unsigned i >= 0) luôn luôn TRUE.
// - Triết lý Bjarne Stroustrup & Google C++ Style Guide:
//   + Dùng 'int' mặc định cho biến đếm/số lượng. Đừng dùng unsigned làm ràng buộc "không âm".
//   + CHỈ DÙNG unsigned cho: Thao tác bit (bit manipulation), Mật mã/Hash (modulo 2^n), hoặc std::size_t.

#include <iostream>

void demoUnsignedPitfalls()
{
    std::cout << "--- 1. CẠM BẪY PHÉP TRỪ QUAY VÒNG DƯỚI (UNDERFLOW) ---\n";
    unsigned int x{ 2 };
    unsigned int y{ 3 };
    std::cout << "2u - 3u = " << x - y << " (Quay vong ve 2^32 - 1)\n\n";

    std::cout << "--- 2. CẠM BẪY SO SÁNH SIGNED VS UNSIGNED ---\n";
    int s{ -1 };
    unsigned int u{ 1 };
    std::cout << "So sanh (-1 < 1u): ";
    if (s < u)
        std::cout << "Dung (-1 < 1)\n";
    else
        std::cout << "SAI! Vi -1 bi ep kieu ngam thanh " << static_cast<unsigned int>(s) << '\n';

    std::cout << "\n--- 3. TRUYỀN SỐ ÂM VÀO THAM SỐ UNSIGNED ---\n";
    auto printSize = [](unsigned int count) {
        std::cout << "So luong nhan duoc: " << count << '\n';
    };
    printSize(-5); // -5 bi ep thanh 4294967291!

    std::cout << "\n--- 4. VÒNG LẶP ĐẾM LÙI AN TOÀN VỚI INT ---\n";
    std::cout << "Dem lui: ";
    for (int i{ 3 }; i >= 0; --i)
    {
        std::cout << i << ' ';
    }
    std::cout << "(Dung chinh xac khi i = -1)\n";
}

int main()
{
    std::cout << "=== BÀI 4.5: UNSIGNED INTEGERS, AND WHY TO AVOID THEM ===\n\n";
    demoUnsignedPitfalls();
    return 0;
}
```

```cpp
//4.6
// ============================================================================
// BÀI 4.6: FIXED-WIDTH INTEGERS AND SIZE_T
// ============================================================================

// 1. TẦNG 1: SỐ NGUYÊN ĐỘ RỘNG CỐ ĐỊNH & SIZE_T (BEGINNER)
// - Vấn đề: int (16/32b), long (32/64b) thay đổi kích thước tùy hệ máy.
// - Giải pháp (<cstdint> C++11): Fixed-width integers đảm bảo kích thước cố định:
//   + Signed: std::int8_t (1B), std::int16_t (2B), std::int32_t (4B), std::int64_t (8B).
//   + Unsigned: std::uint8_t (1B), std::uint16_t (2B), std::uint32_t (4B), std::uint64_t (8B).
// - Cạm bẫy ký tự: int8_t / uint8_t là alias của signed/unsigned char -> std::cout in ra ký tự ASCII ('A', 'B')!
//   + Khắc phục: Dùng +x hoặc static_cast<int>(x).

// 2. TẦNG 2: BẢN CHẤT HỆ THỐNG & C++20 STANDARD (ADVANCED)
// - Bản chất Type Alias: std::int32_t là bí danh (typedef/using) trỏ tới kiểu có sẵn trên máy (như int).
//   + Nếu chip cổ không hỗ trợ kiểu 64-bit, std::int64_t sẽ không tồn tại -> Lỗi biên dịch (Compile Error).
// - std::size_t (<cstddef>):
//   + Là kiểu số nguyên không dấu (unsigned) do compiler định nghĩa để đo kích thước byte đối tượng.
//   + sizeof trả về std::size_t (phụ thuộc độ rộng con trỏ: 4 bytes trên 32-bit, 8 bytes trên 64-bit).
//   + C++20 ([basic.compound] 1.8.2): Kích thước đối tượng không được phép vượt quá max của std::size_t.
// - Fast & Least types: Tránh dùng vì kích thước không đồng nhất giữa các hệ máy.

#include <cstdint>
#include <cstddef>
#include <iostream>

void demoFixedWidthAndSizeT()
{
    std::cout << "--- 1. CÁC KIỂU SỐ NGUYÊN CÓ ĐỘ RỘNG CỐ ĐỊNH (<cstdint>) ---\n";
    std::cout << "sizeof(std::int8_t):   " << sizeof(std::int8_t)   << " byte\n";
    std::cout << "sizeof(std::int16_t):  " << sizeof(std::int16_t)  << " bytes\n";
    std::cout << "sizeof(std::int32_t):  " << sizeof(std::int32_t)  << " bytes\n";
    std::cout << "sizeof(std::int64_t):  " << sizeof(std::int64_t)  << " bytes\n\n";

    std::cout << "--- 2. CẠM BẪY IN KÝ TỰ CỦA int8_t / uint8_t ---\n";
    std::uint8_t letterCode{ 66 };
    std::cout << "In truc tiep std::cout << letterCode: " << letterCode << " (Ra ky tu 'B'!)\n";
    std::cout << "In voi dau (+letterCode):             " << +letterCode << " (Ra so 66!)\n";
    std::cout << "In voi static_cast<int>(letterCode):   " << static_cast<int>(letterCode) << " (Ra so 66!)\n\n";

    std::cout << "--- 3. KIỂU std::size_t (<cstddef>) ---\n";
    int myArray[10]{};
    std::size_t arraySize{ sizeof(myArray) };
    std::cout << "Kich thuoc mang myArray (10 phan tu int): " << arraySize << " bytes\n";
    std::cout << "Kich thuoc std::size_t tren he may nay:    " << sizeof(std::size_t) 
              << " bytes (He thong " << sizeof(std::size_t) * 8 << "-bit)\n";
}

int main()
{
    std::cout << "=== BÀI 4.6: FIXED-WIDTH INTEGERS AND SIZE_T ===\n\n";
    demoFixedWidthAndSizeT();
    return 0;
}
```

```cpp
// ============================================================================
// LESSON 6.7: RELATIONAL OPERATORS & FLOATING POINT COMPARISONS
// ============================================================================
#include <iostream>
#include <iomanip>
#include <algorithm>

// 1. Cạm bẫy IEEE 754: 0.1 + 0.2 != 0.3 do sai số làm tròn số thực nhị phân
std::cout << std::boolalpha;
std::cout << "Kiem tra truc tiep (0.1 + 0.2 == 0.3): " << (0.1 + 0.2 == 0.3) << " (false!)\n";

// 2. So sánh literal cùng kiểu vs khác kiểu (float vs double)
constexpr float f1{ 1.5f };
constexpr double d1{ 1.5 };
std::cout << "1.5f == 1.5 (Luy thua co so 2, exact match): " << (f1 == d1) << " (true)\n";

constexpr float f2{ 6.7f };
constexpr double d2{ 6.7 };
std::cout << "6.7f == 6.7 (Tuan hoan nhi phan, roundoff diff): " << (f2 == d2) << " (false)\n";

// 3. Thuật toán so sánh Epsilon kết hợp (Absolute + Relative) chuẩn công nghiệp
template <typename T>
constexpr T constAbs(T x) {
    return (x < static_cast<T>(0)) ? -x : x;
}

constexpr bool approxEqual(double a, double b, double absEps = 1e-12, double relEps = 1e-8) {
    // Vùng an toàn tuyệt đối cho số sát 0
    if (constAbs(a - b) <= absEps)
        return true;
    // Thuật toán tương đối của Donald Knuth
    double maxVal = (constAbs(a) > constAbs(b)) ? constAbs(a) : constAbs(b);
    return constAbs(a - b) <= (maxVal * relEps);
}

// 4. Kiểm chứng thực tế: Số dư tài khoản ngân hàng sau 10 lần trừ 9.99 USD
double balance{ 100.0 };
for (int i{ 0 }; i < 10; ++i) {
    balance -= 9.99;
}

std::cout << std::setprecision(17);
std::cout << "Balance thuc te: " << balance << '\n';
std::cout << "So sanh balance == 0.1:     " << (balance == 0.1) << " (false!)\n";
std::cout << "So sanh approxEqual(b, 0.1): " << approxEqual(balance, 0.1) << " (true!)\n";

// 5. Kiem chung tai compile-time
constexpr bool compileCheck{ approxEqual(0.1 + 0.2, 0.3) };
std::cout << "Compile-time approxEqual(0.1+0.2, 0.3): " << compileCheck << '\n';
```

```cpp
// ============================================================================
// CHAPTER 6.X: CHAPTER 6 SUMMARY & COMPREHENSIVE QUIZ
// ============================================================================
#include <iostream>
#include <string_view>

// 1. Chained Ternary Operator (?:) trong biểu thức constexpr zero-allocation
constexpr std::string_view getQuantityPhrase(int count) {
    return count < 0 ? "invalid"
         : count == 0 ? "zero"
         : count == 1 ? "a single"
         : count == 2 ? "a pair of"
         : count <= 5 ? "a few"
         : "many";
}

constexpr std::string_view getComponentPluralized(std::string_view singular, int count) {
    return count == 1 ? singular
         : singular == "core" ? "cores"
         : singular;
}

// 2. Compile-time testing với static_assert
static_assert(getQuantityPhrase(-1) == "invalid");
static_assert(getQuantityPhrase(0)  == "zero");
static_assert(getQuantityPhrase(1)  == "a single");
static_assert(getQuantityPhrase(2)  == "a pair of");
static_assert(getQuantityPhrase(4)  == "a few");
static_assert(getQuantityPhrase(10) == "many");
static_assert(getComponentPluralized("core", 1) == "core");
static_assert(getComponentPluralized("core", 4) == "cores");

// 3. De Morgan Guard Clause
auto isAccessDenied = [](bool isRoot, bool hasSshKey, bool isIpWhitelisted, bool isMfaActive) -> bool {
    // Phủ định của isRoot || (hasSshKey && isIpWhitelisted && isMfaActive)
    return !isRoot && (!hasSshKey || !isIpWhitelisted || !isMfaActive);
};

// 4. In kết quả kiểm thử hệ thống
std::cout << "--- CHAPTER 6 COMPREHENSIVE DEMO ---\n";
constexpr int testCounts[] = { -1, 0, 1, 2, 4, 10 };
for (int c : testCounts) {
    std::cout << c << " items -> " << getQuantityPhrase(c) << ' ' 
              << getComponentPluralized("core", c) << '\n';
}

std::cout << "\nAccess Denied Check (Non-root, valid SSH, bad IP): " 
          << std::boolalpha << isAccessDenied(false, true, false, true) << '\n';
```

```cpp
// ============================================================================
// LESSON 7.1: COMPOUND STATEMENTS (BLOCKS) & NESTING DEPTH
// ============================================================================
#include <iostream>
#include <string_view>

// 1. Kỹ thuật Guard Clauses (Early Return) làm phẳng cấu trúc hàm (Nesting Depth <= 2)
auto processPayment = [](bool isAccountActive, double balance, double amount, bool isFraudDetected) -> bool {
    if (!isAccountActive) {
        std::cout << "Declined: Account is locked or inactive.\n";
        return false;
    }

    if (isFraudDetected) {
        std::cout << "Declined: Fraud detected.\n";
        return false;
    }

    if (amount <= 0.0) {
        std::cout << "Declined: Invalid transaction amount.\n";
        return false;
    }

    if (balance < amount) {
        std::cout << "Declined: Insufficient funds.\n";
        return false;
    }

    std::cout << "Transaction approved: " << amount << " USD deducted.\n";
    return true;
};

std::cout << "--- 1. Guard Clauses & Arrow Anti-pattern Refactoring ---\n";
processPayment(true, 500.0, 150.0, false);
processPayment(true, 50.0, 150.0, false);
processPayment(true, 500.0, -10.0, false);
processPayment(true, 500.0, 150.0, true);

// 2. Dangling Else Trap Demo
std::cout << "\n--- 2. Dangling Else Trap Demo ---\n";
auto grantAccess = [](bool isAdmin, bool isHardwareVerified) {
    // Luôn luôn dùng { } để khóa chặt ranh giới câu lệnh
    if (isAdmin) {
        if (isHardwareVerified) {
            std::cout << "ACCESS GRANTED: Full Admin & HW Verified\n";
        }
    } else {
        std::cout << "ACCESS DENIED: Not an Administrator\n";
    }
};
grantAccess(false, true);

// 3. Free-standing block & Scope Lifetime (RAII Mechanism)
std::cout << "\n--- 3. Standalone Block & Lexical Lifetime ---\n";
{
    int localScopedVar{ 999 };
    std::cout << "Inside local standalone block: localScopedVar = " << localScopedVar << '\n';
} // localScopedVar bị giải phóng ngay lập tức khỏi stack tại đây
std::cout << "Exited local standalone block safely.\n";
```

```cpp
// ============================================================================
// LESSON 7.3: LOCAL VARIABLES, STORAGE DURATION & LIFETIME
// ============================================================================
#include <iostream>
#include <utility>

// 1. Pass-by-Reference vs Pass-by-Value: Hoán đổi tọa độ bounding box
auto normalizeBounds = [](int& minVal, int& maxVal) {
    if (minVal > maxVal) {
        int temp{ maxVal }; // temp enters scope & created on stack
        maxVal = minVal;
        minVal = temp;
    } // temp dies & goes out of scope here
};

int top{ 120 }, bottom{ 30 };
int left{ 500 }, right{ 150 };

normalizeBounds(top, bottom);
normalizeBounds(left, right);

std::cout << "--- 1. Bounding Box Normalization ---\n";
std::cout << "Normalized Y-bounds: Top=" << top << ", Bottom=" << bottom << '\n';
std::cout << "Normalized X-bounds: Left=" << left << ", Right=" << right << '\n';

// 2. Minh chứng: In-Lifetime nhưng Out-of-Scope
auto childWorker = [](int id) {
    // Biến 'parentState' ở scope ngoài hoàn toàn OUT-OF-SCOPE ở đây,
    // nhưng nó vẫn đang IN-LIFETIME nằm trên stack frame của caller!
    std::cout << "[Worker Frame] Processing task ID: " << id << '\n';
};

int parentState{ 9999 };
childWorker(parentState);

// 3. Tối thiểu hóa scope: Không rò rỉ biến tính toán tạm
std::cout << "\n--- 2. Scope Isolation in Conditional Block ---\n";
{
    int baseScore{ 80 };
    if (baseScore >= 50) {
        // Biến bonus chỉ tồn tại khi thỏa mãn điều kiện
        int bonus{ 15 };
        std::cout << "Final Score with Bonus: " << baseScore + bonus << '\n';
    } // bonus bị hủy ngay lập tức, không làm ô nhiễm scope bên dưới
}
```

```cpp
// ============================================================================
// LESSON 7.4: INTRODUCTION TO GLOBAL VARIABLES & STATIC DURATION
// ============================================================================
#include <iostream>

// 1. Constexpr Global Constants (An toàn kiến trúc, zero runtime overhead)
namespace ServerConfig {
    constexpr int maxConnections{ 1000 };
    constexpr int serverPort{ 8080 };
}

// 2. Global State có kiểm soát trong Namespace với tiền tố g_
namespace Telemetry {
    // Biến toàn cục có Static Storage Duration (Zero-initialized trong .bss)
    int g_activeConnections{ 0 };

    bool acceptConnection() {
        if (g_activeConnections < ServerConfig::maxConnections) {
            ++g_activeConnections;
            std::cout << "[Accepted] Port: " << ServerConfig::serverPort 
                      << " | Active: " << g_activeConnections << '\n';
            return true;
        }
        std::cout << "[Rejected] Server Full!\n";
        return false;
    }

    void closeConnection() {
        if (g_activeConnections > 0) {
            --g_activeConnections;
            std::cout << "[Closed] Active: " << g_activeConnections << '\n';
        }
    }
}

// 3. Thực thi kiểm thử luồng State Persistence
std::cout << "--- 1. Telemetry Server Connection Simulation ---\n";
Telemetry::acceptConnection();
Telemetry::acceptConnection();
Telemetry::acceptConnection();
Telemetry::closeConnection();

std::cout << "Final Active Connections on Server: " 
          << Telemetry::g_activeConnections << '\n';

// 4. Minh chứng Zero-Initialization vs Static Storage
namespace MemDemo {
    int g_zeroDefault; // Không gán giá trị -> tự động = 0 (Static duration)
}
std::cout << "\n--- 2. Static Storage Zero-Initialization ---\n";
std::cout << "MemDemo::g_zeroDefault = " << MemDemo::g_zeroDefault << " (Guaranteed 0!)\n";
```

```cpp
// ============================================================================
// LESSON 7.5: VARIABLE SHADOWING (NAME HIDING) & UNARY SCOPE RESOLUTION
// ============================================================================
#include <iostream>

// 1. Biến toàn cục (Global Scope)
int g_systemConfig{ 500 };

// 2. Minh họa Local Shadowing & Global Shadowing
void demonstrateShadowing() {
    std::cout << "--- 1. Local Shadowing & Point of Declaration ---\n";
    int balance{ 100 }; // Outer local variable

    {
        // Trước khi biến con xuất hiện, balance vẫn trỏ tới outer variable
        std::cout << "Before inner declaration: balance = " << balance << " (Outer)\n";

        // Khai báo inner balance -> Bắt đầu che khuất (Shadowing) outer balance
        int balance{ 20 }; 
        balance += 30; // Thao tác chỉ tác động lên inner balance

        std::cout << "Inside nested block:      balance = " << balance << " (Inner)\n";
        // Chú ý: Không có bất kỳ cú pháp nào trong C++ để đọc outer balance (100) tại đây!
    }

    // Ra khỏi nested block: inner balance bị hủy, outer balance nguyên vẹn
    std::cout << "After nested block exits: balance = " << balance << " (Outer restored!)\n";

    std::cout << "\n--- 2. Global Shadowing & Unary Scope Operator (::) ---\n";
    int g_systemConfig{ 42 }; // Local variable che khuất biến toàn cục g_systemConfig

    std::cout << "Unqualified lookup:       g_systemConfig = " << g_systemConfig 
              << " (Local Shadow)\n";
    std::cout << "Unary scope resolution: ::g_systemConfig = " << ::g_systemConfig 
              << " (Global Namespace)\n";

    // Thay đổi biến toàn cục một cách tường minh qua ::
    ::g_systemConfig += 100;
    std::cout << "After explicit :: mutation: ::g_systemConfig = " << ::g_systemConfig << '\n';
}

demonstrateShadowing();
```

```cpp
// ============================================================================
// LESSON 7.6: INTERNAL LINKAGE & TRANSLATION UNIT ENCAPSULATION
// ============================================================================
#include <iostream>

// 1. Biến toàn cục có Internal Linkage qua từ khóa `static`
static int g_internalCounter{ 10 }; // Chỉ truy cập được trong TU hiện tại

// 2. Biến toàn cục const / constexpr mặc định có Internal Linkage
const int g_defaultInternalConst{ 100 };       // Mặc định internal linkage
constexpr double g_defaultInternalFactor{ 1.5 }; // Mặc định internal linkage

// 3. Hàm nội bộ (Private helper function) với Internal Linkage
static int computeChecksum(int data) {
    return (data * 31) ^ 0xAA;
}

// 4. Chuẩn Modern C++: Unnamed Namespace (Anonymous Namespace)
// Bao bọc cả biến, hàm và Kiểu dữ liệu (struct/class/enum) trong Internal Linkage
namespace {
    int g_anonymousAuditId{ 505 };

    struct ModuleMetadata {
        int version{ 1 };
        const char* name{ "SecurityCore" };
    };

    void logModuleStatus() {
        ModuleMetadata meta;
        std::cout << "[Internal Module] " << meta.name << " v" << meta.version 
                  << " | Audit ID: " << g_anonymousAuditId << '\n';
    }
}

// 5. Kiểm thử thực thi
std::cout << "--- 1. Static Global & Const Defaults ---\n";
std::cout << "Static Counter:       " << g_internalCounter << '\n';
std::cout << "Const Default:        " << g_defaultInternalConst << '\n';
std::cout << "Constexpr Factor:     " << g_defaultInternalFactor << '\n';
std::cout << "Computed Checksum:    " << computeChecksum(42) << '\n';

std::cout << "\n--- 2. Modern Unnamed Namespace Encapsulation ---\n";
logModuleStatus();
```
